```python
segment_df = df[
    [
        'Tenure Months',
        'Monthly Charges',
        'Total Charges',
        'CLTV'
    ]
]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_data = scaler.fit_transform(segment_df)
```

```python
from sklearn.cluster import KMeans

wcss = []

for i in range(1,11):
    
    kmeans = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(scaled_data)
    
    wcss.append(kmeans.inertia_)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(
    range(1,11),
    wcss,
    marker='o'
)

plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')

plt.title('Elbow Method')

plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

segment_df = df[
    [
        'Tenure Months',
        'Monthly Charges',
        'Total Charges',
        'CLTV'
    ]
]

scaler = StandardScaler()

scaled_data = scaler.fit_transform(segment_df)

wcss = []

for i in range(1,11):
    
    kmeans = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(scaled_data)
    
    wcss.append(kmeans.inertia_)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

df['Cluster'] = kmeans.fit_predict(scaled_data)
```

```python
df['Cluster'].value_counts()
```

```python
cluster_summary = df.groupby('Cluster')[
    [
        'Tenure Months',
        'Monthly Charges',
        'Total Charges',
        'CLTV'
    ]
].mean()

cluster_summary
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

df['Cluster'] = kmeans.fit_predict(scaled_data)

df['Cluster'].value_counts()

cluster_summary = df.groupby('Cluster')[
    [
        'Tenure Months',
        'Monthly Charges',
        'Total Charges',
        'CLTV'
    ]
].mean()

cluster_summary
```

```python
centers = scaler.inverse_transform(
    kmeans.cluster_centers_
)

centers
```

```python
plt.figure(figsize=(12,7))

sns.scatterplot(
    x='Monthly Charges',
    y='CLTV',
    hue='Cluster',
    data=df,
    palette='tab10',
    s=80
)

plt.scatter(
    centers[:,1],
    centers[:,3],
    s=500,
    c='black',
    marker='X',
    label='Centroids'
)

plt.legend()
plt.show()
```

```python
cluster_names = {
    0:'VIP Customers',
    1:'Loyal Customers',
    2:'At-Risk Customers',
    3:'New Customers'
}

df['Segment'] = df['Cluster'].map(cluster_names)
```
